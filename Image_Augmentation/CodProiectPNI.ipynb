## Lucrul cu fisiere si prelularea imaginilor 

```python
from tkinter import Tk
from tkinter.filedialog import askopenfilename

imgFileName = askopenfilename(title="Alege imaginea")
print("Ai selectat:", imgFileName)
```

```python
import matplotlib.image as mpimg
img = mpimg.imread(imgFileName)  
plt.figure()
plt.imshow(img)
plt.axis('off')
plt.show()
```
